## Câu hỏi 3: Trích xuất các dữ liệu trong bảng và làm sạch

### Câu hỏi 3.1: lấy ra hết các hàng trong bảng
Gợi ý: đưa tham số 'tr' vào soup.find_all()

```python
#### YOUR CODE HERE ####

#### END YOUR CODE ####
rows = soup.find_all("tr")
# Print the first 10 rows for sanity check
print(rows[:10])
```

```python
for row in rows:
    row_td = row.find_all('td')
print(row_td)
type(row_td)
```

```python
str_cells = str(row_td)
cleantext = BeautifulSoup(str_cells, "lxml").get_text()
print(cleantext)
```

### Câu hỏi 3.2: làm sạch dữ liệu trong bảng
Đến đây ta đã có dữ liệu của các hàng trong bảng, được lưu ở biến rows. Giờ công việc: sau khi đã lấy dữ liệu cells, hoàn thiện chương trình sau để loại bỏ cac ký tự "\r", "\n" trong dữ liệu cells

```python
import re

list_rows = []
for row in rows:
    cells = row.find_all('td')
    str_cells = str(cells)
    #### YOUR CODE HERE ####
    str_cells=re.sub('\r','',str_cells)
    str_cells=re.sub('\n','',str_cells)
    str_cells = re.sub(r'\s+', ' ', str_cells).strip()
    #### END YOUR CODE ####
    clean = re.compile('<.*?>')
    clean2 = (re.sub(clean,'',str_cells))
    list_rows.append(clean2)
print(list_rows)
type(clean2)
```
